```python
import torch
import numpy as np

def sinusoidal_positional_encoding(seq_len, d_model):
    positions = np.arange(seq_len)[:, np.newaxis]
    div_term = np.exp(np.arange(0, d_model, 2) * -(np.log(10000.0) / d_model))
    
    pos_enc = np.zeros((seq_len, d_model))
    pos_enc[:, 0::2] = np.sin(positions * div_term)
    pos_enc[:, 1::2] = np.cos(positions * div_term)
    
    return torch.tensor(pos_enc, dtype=torch.float32)

def learnable_positional_encoding(seq_len, d_model):
    return torch.nn.Parameter(torch.randn(seq_len, d_model))

# Example with sequence of 6 tokens and embedding size of 8
seq_len = 6
embedding_size = 8

sinusoidal_pe = sinusoidal_positional_encoding(seq_len, embedding_size)
learnable_pe = learnable_positional_encoding(seq_len, embedding_size)

print("Sinusoidal Positional Encoding:")
print(sinusoidal_pe)

print("\nLearnable Positional Encoding (Randomly Initialized):")
print(learnable_pe)
```
