```python
import numpy as np
import nltk
from nltk.tokenize import word_tokenize

# Ensure NLTK tokenizer is ready
nltk.download('punkt')

# Step 1: Tokenize the sentence
sentence = "The quick brown fox jumped over the lazy dog."
tokens = word_tokenize(sentence)
n = len(tokens)  # Number of tokens
d_k = 8  # Embedding size (for simplicity, we use 8)

# Step 2: Create random embeddings for each token
np.random.seed(42)
embeddings = np.random.rand(n, d_k)

# Step 3: Initialize random Q, K, V matrices
W_q = np.random.rand(d_k, d_k)
W_k = np.random.rand(d_k, d_k)
W_v = np.random.rand(d_k, d_k)

Q = embeddings @ W_q
K = embeddings @ W_k
V = embeddings @ W_v

# Step 4: Compute Attention Scores
scores = Q @ K.T / np.sqrt(d_k)  # Scaled dot product
attention_weights = np.exp(scores) / np.sum(np.exp(scores), axis=1, keepdims=True)  # Softmax

# Step 5: Compute final attention output
attention_output = attention_weights @ V

# Display the attention scores
import pandas as pd
attention_df = pd.DataFrame(attention_weights, index=tokens, columns=tokens)
print("\nSelf-Attention Scores:")
print(attention_df)
```
