```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# Load the saved models
generator = tf.keras.models.load_model('generator_final.keras')
discriminator = tf.keras.models.load_model('discriminator_final.keras')

# Verify that the models are loaded successfully
generator.summary()
discriminator.summary()

# Generate fake images using the generator
noise = tf.random.normal([16, 100])  # 16 images, each with 100 noise dimension
generated_images = generator(noise, training=False)

# Display the generated images
plt.figure(figsize=(4, 4))
for i in range(generated_images.shape[0]):
    plt.subplot(4, 4, i + 1)
    plt.imshow(generated_images[i, :, :, 0] * 127.5 + 127.5, cmap='gray')
    plt.axis('off')
plt.show()

# Classify the generated images using the trained discriminator
decision = discriminator(generated_images)

# Print the discriminator's decision (real or fake) for each image
print("Discriminator's decision on generated images:")
for i, output in enumerate(decision.numpy()):
    print(f"Image {i+1}: {'Real' if output > 0 else 'Fake'} (Discriminator Output: {output[0]:.4f})")
```
